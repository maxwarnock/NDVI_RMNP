## Introduction

Rocky Mountain National Park (RMNP) in Colorado is known for its impressive mountains, forests, alpine landscapes, and wildlife. While there are many important ecological features in the park, conifer forests are especially important for a variety of reasons. Forests provide clean water and air, shade, and important habitat for many birds and animals [(Colorado State Forest Service)](https://csfs.colostate.edu/forests-trees/). However, in recent decades, these trees have been under threat from bark beetles and wildfires. To analyze the impact of these threats, I will be using the [Normalized Difference Vegetation Index (NDVI)](https://www.earthdata.nasa.gov/topics/land-surface/normalized-difference-vegetation-index-ndvi) from 2000 to 2024. This index measures "greenness" of vegetation using remotely sensed data, which helps to quantify vegetation health.

## Bark Beetles in Rocky Mountain National Park

According to the National Park Service, there are 17 native species of bark beetles known to occur in the park, and for decades, periodic outbreaks were relatively common but of small scale. Bark beetle populations are affected by winter temperatures, with consistent periods of low temperatures needed to kill off beetles and prevent their populations from growing too rapidly. However, warmer average temperatures in the park in recent decades have allowed beetle populations to dramatically increase, and the recent outbreak has been more severe than any before it [(Mountain Pine Beetle. NPS)](https://www.nps.gov/romo/learn/nature/mtn_pine_beetle_background.htm). Across Colorado, mountain pine beetles affected 3.4 million acres of forest from the late 1990s through 2013 [(Mountain Pine Beetle. CSFS)](https://csfs.colostate.edu/forest-management/common-forest-insects-diseases/mountain-pine-beetle/). In the park, several species of bark beetles are killing lodgepole pine, ponderosa pine, limber pine, Engelmann spruce, subalpine fir, and Colorado blue spruce [(Mountain Pine Beetle. NPS)](https://www.nps.gov/romo/learn/nature/mtn_pine_beetle_background.htm).

## Wildfires in Rocky Mountain National Park

Fires are a natural part of forest ecosystems. However, the park has experienced several large fires in recent years [(Fire History. NPS)](https://www.nps.gov/romo/learn/fire-history.htm). In 2012, the Fern Lake Fire burned nearly 3,500 acres in the park, with beetle-killed trees adding to the danger for firefighters. In 2020, the Cameron Peak and East Troublesome Fires burned roughly 30,000 acres within the park [(Fire History. NPS)](https://www.nps.gov/romo/learn/fire-history.htm).

## Timeframe

To analyze the impacts of bark beetles and wildfires, I will be using NDVI data from 2000 to 2024. This timeframe will encompass most of the recent bark beetle outbreak, and it will also capture the more recent wildfires from 2012 and 2020, as well as any recent regrowth from these impacts. Additionally, I'm only interested in the vegetation health of evergreen conifers (pine, spruce, and fir), so I will be downloading winter data to account for this. The MODIS 16-day composites that cover January run from mid-December through January. At that time of year, the only green vegetation cover should be evergreen conifers, with other vegetation either covered in snow or dormant.

## Data

I will be using Moderate Resolution Imaging Spectroradiometer (MODIS) NDVI data (the MOD13Q1 version 6.1 product; Didan, 2021) downloaded from [NASA's Earthdata catalog](https://www.earthdata.nasa.gov/). Specifically, I will be downloading via NASA's Application for Extracting and Exploring Analysis Ready Samples [(AppEEARS)](https://appeears.earthdatacloud.nasa.gov/explore) using the earthpy package. The park boundary comes from OpenStreetMap.

## Calculating a Difference Map

The map below shows change in NDVI values for the two date slices. I subtracted the earlier date slice (2000-2012) from the more recent date slice (2012-2024). The result shows large areas of pink which indicates a decrease in vegetation health over the time period. There are also some green areas which indicate an increase in vegetation health over the time period. Overall, it's not surprising to see a decrease in NDVI values due to the impacts from bark beetles and wildfires, as described above.

## Conclusion

The difference map, the yearly averages, and the trend line all indicate a clear decrease in NDVI values in Rocky Mountain National Park over the 24-year study period. In my data preparation and methods, I was careful to download data that would give results for evergreen conifers specifically, by downloading data from mid-December through January when other vegetation is dormant. This should remove potential errors that could have otherwise occurred if I used data from the summer for example.

The decrease in NDVI values can be explained by the bark beetle outbreak, which affected Colorado's forests from the late 1990s through 2013 [(Mountain Pine Beetle. CSFS)](https://csfs.colostate.edu/forest-management/common-forest-insects-diseases/mountain-pine-beetle/). Additionally, significant wildfires in the park during 2012 and 2020 are another factor that is likely causing the decrease in NDVI values [(Fire History. NPS)](https://www.nps.gov/romo/learn/fire-history.htm). One limitation is that winter NDVI in a high-elevation park also changes with snow cover, so some of the year-to-year variation may reflect snow rather than tree health.

This analysis shows the usefulness of using NDVI data to quantify changes in vegetation health. This analysis could be taken further by using other datasets, such as wildfire maps, or beetle kill maps (if they exist) to cross check the results found above. Additionally, knowing vegetation health trends is an important tool for conservation efforts and climate change understanding and mitigation.
